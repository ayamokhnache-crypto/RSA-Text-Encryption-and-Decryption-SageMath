```python
# a function to code the message in numbers 
def coding_the_message (message, code_array, letter_array):
    message_array=list(message)                             # puting each character of the message in an array 
    message_in_number =[]
    for i in range (0,len(message_array)):                  # looking for the position of the letter in the letters array 
        for j in range (0,len(letter_array)):
            if message_array[i]==letter_array[j]:           # finding its appropriate number 
                message_in_number.append(code_array[j])     
    return message_in_number          
```

```python
#function to calculate the modulo power of a base
def power(base, expo, m):
    result = 1
    # Update base if it is greater than or equal to mod
    base = base % m  
    
    while expo > 0:
        # If expo is odd, multiply base with result
        if expo % 2 == 1:
            result = (result * base) % m
        
        # Square the base and divide expo by 2
        base = (base * base) % m
        expo //= 2 # Integer division
        
    return result
```

```python
# checking if a number is prime using the Miller_Rabin test
def is_prime_Miller_Rabin(p):
    if p==2:
        return True
    if p<2 or p % 2 == 0 :
        return False
    
    # writing n-1=2**s*d such that d should be odd
    d = p - 1
    s = 0                 # counter for the power of 2 
    while d % 2 == 0:     # to check d is odd if not continue 
        d //= 2
        s += 1

    # perform k rounds
    k=40
    for i in range(k):
        a = ZZ.random_element(2, p - 2)      # we choose a random base a 
        x = power(a, d, p)

        if x == 1 or x == p - 1:
            continue                         # it means # probably prime so we do not complete the next steps we directly redo the process for k+1

        not_prime  = True                    # else it is not prime and we do the square repeatedly
        for j in range(s - 1):                
            x = power(x, 2, p)
            if x == p - 1:
                not_prime = False            # probably prime so we beak directely and we try for k+1
                break

        if not_prime:
            return False         # not prime    

    return True                  # if it passes the loop then it is prime 
```

```python
#function to convert a decimal number to a binary repreentation

def int_to_bin(n):
    bin_rep=[]                                         
    # this formula for calculating the number of bits required to represent a decimal number in binary
    number_of_bits = n.bit_length()
    for i in range(1,number_of_bits + 1):            #The sucsisive division of n over the powers of 2 
            if(n/(2**(number_of_bits-i)) <1):           #Cheking each time if the division of n over the power of two gives a number smaller than 1
                bin_rep.append(0)                       # we append 1
            else:                                        #if it is bigger we add 1 in the corresponding position 
                bin_rep.append(1)
                n = n - 2**(number_of_bits-i)            #Each time we divide and add one , we update n by subtracting that value
   
    return bin_rep 
```

```python
#function to calculate the exponent e
import random
def calculating_e(phi):
    bin_rep = int_to_bin(phi)                            
    k = len(bin_rep)                                     
    valid_e = []                                         
    #in practice e is chosed to be the number with small hamming weight (to make the encryption easier)
    #the smallest hamming weight in RSA is 2 , else(the smallest hamming weight = 1) we have just one in one position of power of 2:
    # if the power is 0 means e = 1 , rejected ( e must be strictly greater then 1)
    # if the power is not zero means e is ever so gcd(e,phi) cannot be 1 , rejected
    # the hamming weight is 2 ,  we have 1 in exactly two positions :
    # in the first position (LSB): since e is always odd
    #anthor position
    for i in range(0,k):                                  #stopping at k where k is the number of bits represented phi(since e < phi)
        e = 2**i + 1                                      # this is exactly the formula of e according to the above explanation (Fermat exponent)
        if gcd(phi, e) == 1:                              #ensuring that e is coprime with phi
            if e < phi:                                   #ensuring that e is less than phi
               valid_e.append(e)                          
    #since e = 65537 is the commenly used exponent , we check if it is in the array we selected as the exponent 
    if 65537 in valid_e:
        e = 65537
    # if not , we select a random one from the array of valid values of e
    else:
         e = random.choice(valid_e)
    return e
```

```python
#function to calculate th egreatest commen divisor of teo numbers using Euclidean algorithm 
def greatest_commen_divisor(a,b):
    # solve the equation  ax + by = gcd(a,b)
    #insialization 
    x0 ,x1 =1  , 0 
    y0 ,y1 = 0 , 1
    r0 , r1 = a ,b 
    while r1 != 0:                                                    #we continue until the reminder is 0
        q = r0 // r1                                                  # updating coefficients
        r0 , r1 = r1 ,r0 - q*r1
        x0 ,x1 = x1  , x0 - q*x1
        y0 ,y1 = y1 ,y0 - q*y1
    return r0 , x0 , y0                                               #return gcd , x,y
```

```python
#function to find the modulo inverse of e
def find_d(e, phi):
    gcd , d_first , l = greatest_commen_divisor(e, phi)     #solving the equation e.d - k.phi = 1
    d_correct = d_first % phi                               # wroting d modulo phi
    if ((e * d_correct)%phi == 1):                          # checking the correctness
        return d_correct
```

```python
# finding n 
def RSA_n (message_in_number):
    max_number=max(message_in_number)                    # finding the max number that code a letter in our message 
    n=0                                                  # to be able to use n as a condition in while
    while(n<=max_number):                                
        p=0
        while not is_prime_Miller_Rabin(p):                           # choosing large prime numbers p and q 
            p = ZZ.random_element(2**1023,2**2000)

        q=0
        while not is_prime_Miller_Rabin(q) or p==q :
            q = ZZ.random_element(10, 1000)
        n=p*q
    return n ,p ,q
```

```python
# RSA method for encrypting the message
def RSA_method_encrypted (block_number,n, p, q):
    phi=(p-1)*(q-1)
    e = calculating_e(phi)
    d = find_d(e, phi)
    
    encrypted = [power_mod(num, e, n) for num in block_number]
    
    return encrypted , d
```

```python
#RSA method for decrypting the message using CRT
def RSA_method_decrypted (encrypted, d, n,p,q):
    dp = d % (p - 1)
    dq = d % (q - 1)
    gcd, q_inv, _ = greatest_commen_divisor(q, p)
    q_inv = q_inv % p
    decrypted_list = []
    
    for c in encrypted:
        m1 = power_mod(c, dp, p)
        m2 = power_mod(c, dq, q)
        
        h = (q_inv * (m1 - m2)) % p
        m = m2 + h * q
        
        decrypted_list.append(m)
        
    return decrypted_list
```

```python
# a function to go back to the message in letters 
def decoding_the_message (RSA_decrypted, code_array, letter_array):
    back_to_message =[]
    for i in range (0,len(RSA_decrypted)):                 
        for j in range (0,len(code_array)):                
            if RSA_decrypted[i]==code_array[j]:              # looking for the position of the number in the code array
                back_to_message.append(letter_array[j])      # insert the letter of the same position 
    return back_to_message 
```
